**✅ 1 – Two Sum**

````python
def twosum(nums,target):
  sum = {}
  for i in range(len(nums)):
    for j in range(i+1,len(nums)):
      if nums[i]+nums[j] == target:
        sum = i,j
        print(sum)
nums = [2,7,11,15]
target = 9
twosum(nums,target)
````

````python
def twosum(nums,target):
  sum = {}
  for i, num in enumerate(nums):
    t = target - num
    if t in sum:
      return [sum[t],i]
    sum[num] = i

nums = [2,7,11,15]
target = 9
twosum(nums,target)
````

---

# Intuition

Brute force approach me hum har pair check karte hain (nested loop), lekin uska time complexity (O(n^2)) hota hai.

Better approach ke liye hume fast lookup chahiye.
Dictionary (hash map) O(1) lookup provide karta hai.

Idea yeh hai:

* Har number ke liye check karo ki uska complement (`target - num`) pehle array me aaya hai ya nahi.
* Agar aaya hai → answer mil gaya.
* Nahi aaya → current number ko dictionary me store kar do.

---

# Approach

1. Ek empty dictionary `seen` create karo.
2. Array ko `enumerate()` se iterate karo.
3. Har element ke liye:

   * `complement = target - num`
   * Agar `complement` dictionary me hai:

     * Return `[seen[complement], i]`
   * Otherwise:

     * Current number aur uska index dictionary me store kar do.
4. Guaranteed ek solution hota hai, isliye return mil jayega.

---

# Complexity

* **Time complexity:**
  (O(n))
  Kyunki hum array ko sirf ek baar traverse kar rahe hain aur dictionary lookup (O(1)) hota hai.

* **Space complexity:**
  (O(n))
  Worst case me saare elements dictionary me store ho sakte hain.

---

# Code

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i
```
